```python
import simpy
import random

# Create an empty SimPy environment.
env = simpy.Environment()

print(f"Empty SimPy environment created. Current time: {env.now}")
```

```python
def tiny_process(env):
    print(f"start at {env.now}")
    yield env.timeout(5)
    print(f"end at {env.now}")

process = env.process(tiny_process(env))
env.run()
```

```python
## paramaters
num_of_households = 10
```

```python
#create a list of households. 1 represents returned census, 0 reprsents not returned. 
hh_list = []
for i in range (num_of_households):
    hh_list.append(random.randint(0, 1))
print(hh_list)

```

```python
def update_households_at_time_10(env, hh_list):
    yield env.timeout(max(0, 10 - env.now))
    for index, value in enumerate(hh_list):
        if value == 0:
            hh_list[index] = 1
    print(f"At time {env.now}, hh_list is now: {hh_list}")


print(f"Before event at time {env.now}: {hh_list}")
env.process(update_households_at_time_10(env, hh_list))
env.run()
```
